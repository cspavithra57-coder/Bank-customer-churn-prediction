```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

print("Libraries imported successfully.")
```

```python
from google.colab import files
from io import BytesIO

uploaded = files.upload()

# Select the uploaded CSV file
csv_files = [name for name in uploaded.keys() if name.lower().endswith(".csv")]

if not csv_files:
    raise FileNotFoundError("Please upload the European_Bank.csv file.")

file_name = csv_files[0]

data = pd.read_csv(BytesIO(uploaded[file_name]))

print("Dataset loaded successfully.")
print("Dataset shape:", data.shape)

display(data.head())
```

```python
print("Dataset Information:")
data.info()

print("\nNumber of rows and columns:")
print(data.shape)

print("\nColumn Names:")
print(data.columns.tolist())

print("\nFirst 5 Records:")
display(data.head())

print("\nLast 5 Records:")
display(data.tail())

print("\nStatistical Summary:")
display(data.describe(include="all").T)
```

```python
print("Missing Values in Each Column:")
display(data.isnull().sum().to_frame("Missing Values"))

print("\nTotal Missing Values:")
print(data.isnull().sum().sum())

print("\nNumber of Duplicate Rows:")
print(data.duplicated().sum())
```

```python
# Remove duplicate rows
data = data.drop_duplicates().copy()

# Check required columns
required_columns = [
    "Year",
    "CustomerId",
    "Surname",
    "CreditScore",
    "Geography",
    "Gender",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "HasCrCard",
    "IsActiveMember",
    "EstimatedSalary",
    "Exited"
]

missing_columns = [
    col for col in required_columns if col not in data.columns
]

if missing_columns:
    raise ValueError(f"Required columns are missing: {missing_columns}")

# Confirm target values
print("Target variable values:")
print(data["Exited"].value_counts(dropna=False))

# Remove rows with missing target values
data = data.dropna(subset=["Exited"]).copy()

# Ensure the target is numeric
data["Exited"] = pd.to_numeric(data["Exited"], errors="raise")

# Remove irrelevant columns
data.drop(
    columns=["Year", "CustomerId", "Surname"],
    inplace=True
)

print("\nDataset after removing duplicate rows and irrelevant columns:")
print("Shape:", data.shape)

display(data.head())
```

```python
# Identify numeric and categorical columns
numeric_columns = data.select_dtypes(include=np.number).columns.tolist()
categorical_columns = data.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

# Fill missing numeric values using the median
for col in numeric_columns:
    if col != "Exited" and data[col].isnull().any():
        data[col] = data[col].fillna(data[col].median())

# Fill missing categorical values using the mode
for col in categorical_columns:
    if data[col].isnull().any():
        mode_values = data[col].mode()

        if not mode_values.empty:
            data[col] = data[col].fillna(mode_values.iloc[0])

print("Missing values after handling:")
display(data.isnull().sum().to_frame("Remaining Missing Values"))
```

```python
# Replace zero salary values with NaN to avoid division by zero
salary_denominator = data["EstimatedSalary"].replace(0, np.nan)

data["Balance_Salary_Ratio"] = (
    data["Balance"] / salary_denominator
)

# Handle undefined or infinite ratios
data["Balance_Salary_Ratio"] = (
    data["Balance_Salary_Ratio"]
    .replace([np.inf, -np.inf], np.nan)
)

# Fill undefined ratios with 0 for a usable model input
data["Balance_Salary_Ratio"] = (
    data["Balance_Salary_Ratio"].fillna(0)
)

print("Balance-Salary Ratio created successfully.")

display(
    data[
        ["Balance", "EstimatedSalary", "Balance_Salary_Ratio"]
    ].head(10)
)
```

```python
data["Product_Density"] = (
    data["NumOfProducts"] / (data["Tenure"] + 1)
)

print("Product Density created successfully.")

display(
    data[
        ["NumOfProducts", "Tenure", "Product_Density"]
    ].head(10)
)
```

```python
data["Engagement_Product_Interaction"] = (
    data["IsActiveMember"] * data["NumOfProducts"]
)

print("Engagement-Product Interaction created successfully.")

display(
    data[
        [
            "IsActiveMember",
            "NumOfProducts",
            "Engagement_Product_Interaction"
        ]
    ].head(10)
)
```

```python
data["Age_Tenure_Interaction"] = (
    data["Age"] * data["Tenure"]
)

print("Age-Tenure Interaction created successfully.")

display(
    data[
        ["Age", "Tenure", "Age_Tenure_Interaction"]
    ].head(10)
)
```

```python
engineered_features = [
    "Balance_Salary_Ratio",
    "Product_Density",
    "Engagement_Product_Interaction",
    "Age_Tenure_Interaction"
]

print("Engineered Features:")
display(data[engineered_features].head(10))

print("\nSummary Statistics of Engineered Features:")
display(data[engineered_features].describe().T.round(4))
```

```python
print("Correlation of Engineered Features with Exited:")

correlations = (
    data[engineered_features + ["Exited"]]
    .corr(numeric_only=True)["Exited"]
    .drop("Exited")
    .sort_values(ascending=False)
    .to_frame("Correlation with Exited")
)

display(correlations.round(4))
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=correlations["Correlation with Exited"],
    y=correlations.index
)

plt.title("Correlation of Engineered Features with Customer Churn")
plt.xlabel("Correlation with Exited")
plt.ylabel("Engineered Feature")
plt.axvline(x=0, color="black", linewidth=1)
plt.tight_layout()
plt.show()
```

```python
print("Dataset after feature engineering:")
print("Shape:", data.shape)

print("\nColumn Names:")
print(data.columns.tolist())

print("\nFirst 10 Records:")
display(data.head(10))

print("\nMissing Values:")
display(data.isnull().sum().to_frame("Missing Values"))
```

```python
X = data.drop(columns=["Exited"]).copy()
y = data["Exited"].copy()

print("Input Features Shape Before Encoding:", X.shape)
print("Target Variable Shape:", y.shape)

print("\nTarget Variable Distribution:")
display(
    y.value_counts()
    .sort_index()
    .rename(index={0: "Stayed (0)", 1: "Exited (1)"})
    .to_frame("Number of Customers")
)
```

```python
categorical_features = X.select_dtypes(
    include=["object", "category", "string"]
).columns.tolist()

print("Categorical Features to Encode:")
print(categorical_features)

# One-hot encode categorical variables
X = pd.get_dummies(
    X,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

print("\nEncoding completed successfully.")
print("Final X shape:", X.shape)
print("Final y shape:", y.shape)

print("\nFinal Input Feature Columns:")
print(X.columns.tolist())

display(X.head())
```

```python
assert set(y.unique()).issubset({0, 1}), (
    "The target variable must contain only 0 and 1."
)

# Check matching row counts
assert len(X) == len(y), (
    "X and y must contain the same number of rows."
)

# Check for missing or infinite values in model inputs
assert not X.isnull().any().any(), (
    "Missing values remain in X."
)

assert np.isfinite(X.to_numpy(dtype=float)).all(), (
    "Infinite or non-finite values remain in X."
)

# Check that all model input columns are numeric
assert all(
    pd.api.types.is_numeric_dtype(dtype)
    for dtype in X.dtypes
), "All columns in X should be numeric."

print("All validation checks passed.")
print("Final X shape:", X.shape)
print("Final y shape:", y.shape)
print("Number of input features:", X.shape[1])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, RobustScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    RocCurveDisplay
)

from google.colab import files
import os

file_name = "European_Bank_Feature_Engineered.csv"

if not os.path.exists(file_name):
    uploaded = files.upload()
    file_name = next(
        name for name in uploaded
        if name.endswith(".csv")
    )

data = pd.read_csv(file_name)

print("Dataset shape:", data.shape)
print("Columns:", data.columns.tolist())
display(data.head())
```

```python
X = data.drop(columns=["Exited"]).copy()
y = data["Exited"].astype(int).copy()

# Remove rows with invalid target values, if any
valid_target = y.isin([0, 1])
X = X.loc[valid_target].copy()
y = y.loc[valid_target].copy()

print("Input features:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())

print("\nChurn percentage:")
print((y.value_counts(normalize=True) * 100).round(2))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

print("\nTraining churn rate:", round(y_train.mean() * 100, 2), "%")
print("Testing churn rate:", round(y_test.mean() * 100, 2), "%")
```

```python
categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numeric_features = X.select_dtypes(
    include=np.number
).columns.tolist()

numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", RobustScaler())
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ],
    remainder="drop"
)
print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=5,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    )
}

trained_models = {}
predictions = {}
probabilities = {}

for name, model in models.items():
    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)

    trained_models[name] = pipeline
    predictions[name] = pipeline.predict(X_test)
    probabilities[name] = pipeline.predict_proba(X_test)[:, 1]

    print(name, "training completed.")

print("\nAll four models have been trained.")
```

CH 7

```python
results = []

for name in trained_models:
    y_pred = predictions[name]
    y_prob = probabilities[name]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1-Score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="ROC-AUC",
    ascending=False
).reset_index(drop=True)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)

results_df.to_csv(
    "Chapter_7_Model_Comparison.csv",
    index=False
)
```

```python
metrics_to_plot = [
    "Accuracy", "Precision", "Recall", "F1-Score", "ROC-AUC"
]

results_df.set_index("Model")[metrics_to_plot].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Comparison of Bank Customer Churn Prediction Models")
plt.ylabel("Score")
plt.xlabel("Machine Learning Model")
plt.ylim(0, 1)
plt.xticks(rotation=20, ha="right")
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python
for name in trained_models:
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    cm = confusion_matrix(
        y_test,
        predictions[name],
        labels=[0, 1]
    )

    plt.figure(figsize=(5, 4))
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Stayed (0)", "Churned (1)"],
        yticklabels=["Stayed (0)", "Churned (1)"]
    )

    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.tight_layout()
    plt.show()

    print(classification_report(
        y_test,
        predictions[name],
        labels=[0, 1],
        target_names=["Stayed", "Churned"],
        zero_division=0
    ))
```

```python
plt.figure(figsize=(9, 7))

for name in trained_models:
    RocCurveDisplay.from_predictions(
        y_test,
        probabilities[name],
        name=name,
        ax=plt.gca()
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random classification"
)

plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
best_model_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]

print("Top model by ROC-AUC:", best_model_name)
print(
    "ROC-AUC:",
    round(results_df.iloc[0]["ROC-AUC"], 4)
)

print("\nFull performance results:")
display(results_df)
```

```python
risk_data = X_test.copy()

risk_data["Actual_Exited"] = y_test.to_numpy()

risk_data["Predicted_Churn_Probability"] = (
    best_model.predict_proba(X_test)[:, 1]
)

risk_data["Predicted_Exited"] = (
    best_model.predict(X_test)
)

def assign_risk(probability):
    if probability < 0.30:
        return "Low Risk"
    elif probability < 0.60:
        return "Medium Risk"
    else:
        return "High Risk"

risk_data["Churn_Risk_Category"] = (
    risk_data["Predicted_Churn_Probability"]
    .apply(assign_risk)
)

risk_data = risk_data.sort_values(
    "Predicted_Churn_Probability",
    ascending=False
)

display(risk_data.head(10))

risk_data.to_csv(
    "Chapter_7_Customer_Churn_Risk_Scores.csv",
    index=False
)

print("Customer risk scores saved successfully.")
```

```python
risk_summary = (
    risk_data["Churn_Risk_Category"]
    .value_counts()
    .reindex(
        ["Low Risk", "Medium Risk", "High Risk"],
        fill_value=0
    )
    .rename_axis("Risk Category")
    .reset_index(name="Number of Customers")
)

risk_summary["Percentage"] = (
    risk_summary["Number of Customers"]
    / len(risk_data) * 100
).round(2)

display(risk_summary)

sns.barplot(
    data=risk_summary,
    x="Risk Category",
    y="Number of Customers"
)

plt.title("Customer Distribution by Predicted Churn Risk")
plt.xlabel("Risk Category")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```
